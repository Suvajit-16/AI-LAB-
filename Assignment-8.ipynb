```python
"""Write a python program to implement the minimax algorithm for a two-player game.

1. Implement the Minimax algorithm using recursion.
2. Consider MAX and MIN players.
3. Use the following leaf-node values:

   3, 5, 2, 9, 12, 5, 23, 23

4. Determine the optimal score for the MAX player.
5. Display the final optimal score as output.
import math"""

def minimax(depth, node_index, is_maximizing, values, max_depth):
    # Base case: leaf node
    if depth == max_depth:
        return values[node_index]

    if is_maximizing:
        return max(
            minimax(depth + 1, node_index * 2, False, values, max_depth),
            minimax(depth + 1, node_index * 2 + 1, False, values, max_depth)
        )
    else:
        return min(
            minimax(depth + 1, node_index * 2, True, values, max_depth),
            minimax(depth + 1, node_index * 2 + 1, True, values, max_depth)
        )
# Leaf node values
leaf_values = [3, 5, 2, 9, 12, 5, 23, 23]
tree_depth = int(math.log2(len(leaf_values)))

best_score = minimax(0, 0, True, leaf_values, tree_depth)
# Display result
print("Leaf Node Values:", values)
print("Best possible score:", best_score)
```

```python
#write a python program to implement the Alpha-Beta Puring algorithm for a two-player game. Use the leaf-node values 3, 5, 2, 9, 12, 5, 23, 23. Determine the optimal score for the MAX player and display which branches are pruned during the search.
# Alpha-Beta Pruning Algorithm

def alpha_beta(depth, node_index, maximizing_player, values, alpha, beta, max_depth):

    # Base case: leaf node
    if depth == max_depth:
        return values[node_index]

    if maximizing_player:
        # MAX player
        best = float('-inf')

        for i in range(2):
            value = alpha_beta(
                depth + 1,
                node_index * 2 + i,
                False,
                values,
                alpha,
                beta,
                max_depth
            )

            best = max(best, value)
            alpha = max(alpha, best)

            # Beta pruning
            if beta <= alpha:
                print("Pruned branch at node:", node_index * 2 + i)
                break

        return best

    else:
        # MIN player
        best = float('inf')

        for i in range(2):
            value = alpha_beta(
                depth + 1,
                node_index * 2 + i,
                True,
                values,
                alpha,
                beta,
                max_depth
            )

            best = min(best, value)
            beta = min(beta, best)

            # Alpha pruning
            if beta <= alpha:
                print("Pruned branch at node:", node_index * 2 + i)
                break

        return best


# Leaf-node values
values = [3, 5, 2, 9, 12, 5, 23, 23]

# Calculate maximum depth
max_depth = 3

# Initial Alpha and Beta values
alpha = float('-inf')
beta = float('inf')

print("Leaf Node Values:")
print(values)

print("\nBranches pruned during the search:")

# Call Alpha-Beta Pruning
optimal_score = alpha_beta(
    0,
    0,
    True,
    values,
    alpha,
    beta,
    max_depth
)

# Display the final result
print("\nOptimal score for MAX player:", optimal_score)
```
